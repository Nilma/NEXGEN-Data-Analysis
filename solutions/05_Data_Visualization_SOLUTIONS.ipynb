```python
from pathlib import Path
import pandas as pd
DATA = Path('../data') if Path('../data').exists() else Path('data')

import matplotlib.pyplot as plt
c = pd.read_csv(DATA / "consultants.csv")
a = pd.read_csv(DATA / "assignments.csv")
t = pd.read_csv(DATA / "timesheets.csv")
p = pd.read_csv(DATA / "projects.csv")
w = t.merge(a,on="AssignmentID").merge(c,on="ConsultantID").merge(p,on="ProjectID")
w["Revenue"] = w["Hours"] * w["HourlyRate"]
print(w.groupby("FullName")["Hours"].sum().sort_values(ascending=False))
r = w.groupby("Title")["Revenue"].sum().sort_values(ascending=False)
r.plot.bar(title="Recorded revenue by project",ylabel="Revenue (DKK)")
plt.xticks(rotation=20,ha="right")
plt.tight_layout()
plt.show()
```

Discuss small sample size, recorded time vs total project work, and revenue vs profitability.
