```python
from pathlib import Path
import pandas as pd
DATA = Path('../data') if Path('../data').exists() else Path('data')

df = pd.read_csv(DATA / "consultants_messy_PRACTICE.csv")
before = len(df)
print("Mean before:", df["ExperienceYears"].mean())
df["FullName"] = df["FullName"].str.strip()
df = df.drop_duplicates(subset=["ConsultantID"])
print("Rows removed:", before-len(df))
df["ExperienceYears"] = df["ExperienceYears"].fillna(df["ExperienceYears"].median())
print("Mean after:", df["ExperienceYears"].mean())
```

Median imputation can distort distributions and mask a meaningful missingness pattern.
