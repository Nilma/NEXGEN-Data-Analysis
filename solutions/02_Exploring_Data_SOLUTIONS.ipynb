```python
from pathlib import Path
import pandas as pd
DATA = Path('../data') if Path('../data').exists() else Path('data')

c = pd.read_csv(DATA / "consultants.csv")
t = pd.read_csv(DATA / "timesheets.csv")
print(c[c["ExperienceYears"] < 8])
print(t.sort_values("Hours", ascending=False))
print("Average hours:", t["Hours"].mean())
```

A pandas boolean filter is similar to `WHERE`, but works on data already loaded into Python.
