# Lesson 3 — Clean data
This lesson uses an **intentionally altered practice CSV** (missing experience, whitespace and duplicate ID). The original SQL data and clean CSV are unchanged.

```python
from pathlib import Path
import pandas as pd
DATA = Path('../data') if Path('../data').exists() else Path('data')

df = pd.read_csv(DATA / "consultants_messy_PRACTICE.csv")
df
```

```python
df.isna().sum()
```

```python
df.duplicated(subset=["ConsultantID"]).sum()
```

```python
df["FullName"] = df["FullName"].str.strip()
df = df.drop_duplicates(subset=["ConsultantID"])
df
```

### Missing values: decide, do not guess
For this demonstration, fill the missing experience with the **median**. In real analysis, discuss whether this is appropriate; do not silently change source facts.

```python
df["ExperienceYears"] = df["ExperienceYears"].fillna(df["ExperienceYears"].median())
df
```
