# Lesson 5 — Analyse and visualise NEXGEN
We join the CSV tables in pandas, similar to SQL `JOIN`. The figures represent **sample data only** (4 consultants, 2 projects, 10 timesheets).

```python
from pathlib import Path
import pandas as pd
DATA = Path('../data') if Path('../data').exists() else Path('data')

import matplotlib.pyplot as plt
consultants = pd.read_csv(DATA / "consultants.csv")
assignments = pd.read_csv(DATA / "assignments.csv")
timesheets = pd.read_csv(DATA / "timesheets.csv")
projects = pd.read_csv(DATA / "projects.csv")
```

```python
work = timesheets.merge(assignments, on="AssignmentID", validate="many_to_one").merge(consultants, on="ConsultantID", validate="many_to_one").merge(projects, on="ProjectID", validate="many_to_one")
work["Revenue"] = work["Hours"] * work["HourlyRate"]
work.head()
```

```python
by_consultant = work.groupby("FullName", as_index=False)["Hours"].sum().sort_values("Hours", ascending=False)
by_consultant
```

```python
ax = by_consultant.plot.bar(x="FullName", y="Hours", legend=False, title="Recorded hours by consultant")
ax.set_ylabel("Hours")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

```python
by_project = work.groupby(["ProjectID", "Title"], as_index=False).agg(TotalHours=("Hours", "sum"), Revenue=("Revenue", "sum"), Budget=("Budget", "first"))
by_project["RevenuePctOfBudget"] = 100 * by_project["Revenue"] / by_project["Budget"]
by_project
```

### Interpretation
Revenue is **hours × hourly rate**, not profit. The budget comparison is illustrative: budget and recorded revenue are different measures. Missing time entries and other costs may change the story.
