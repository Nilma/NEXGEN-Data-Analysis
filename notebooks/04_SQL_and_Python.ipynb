# Lesson 4 — Connect Jupyter to MySQL
**Prerequisite:** Run `database/SQLCreateNexgen.sql` then `SQLInsertNexgen.sql` in MySQL Workbench. **Warning:** the supplied CREATE script begins with `DROP DATABASE IF EXISTS nexgen` and deletes any existing `nexgen` database. Review it before running.

Start MySQL server separately; it does **not** run from your Python notebook. Credentials are prompted, not saved.

```python
import os
from getpass import getpass
import pandas as pd
from sqlalchemy import create_engine, text
from sqlalchemy.engine import URL
```

```python
user = input("MySQL user [root]: ").strip() or "root"
password = getpass("MySQL password: ")
port = int(input("MySQL port [3306]: ").strip() or "3306")
url = URL.create("mysql+pymysql", username=user, password=password, host="127.0.0.1", port=port, database="nexgen")
engine = create_engine(url, pool_pre_ping=True)
```

```python
with engine.connect() as conn:
    df = pd.read_sql(text("SELECT ConsultantID, FullName, ExperienceYears FROM Consultant"), conn)
df
```

```python
with engine.connect() as conn:
    billing = pd.read_sql(text("SELECT * FROM v_ProjectBilling ORDER BY ProjectID"), conn)
billing
```
